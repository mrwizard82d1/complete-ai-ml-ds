## Dot product example (nut butter sales)

```python
# Number of jars sold of each type
sales_amounts = rng.integers(20, size=(5, 3))
sales_amounts
```

```python
# Create weekly_sales DataFrame
weekly_sales = pd.DataFrame(sales_amounts, 
                            index=['Mon', 'Tue', 'Wed', 'Thu', 'Fri'],
                            columns=['Almond butter', 'Peanut butter', 'Cashew butter'])
weekly_sales
```

```python
# Create prices array
prices = np.array([10, 8, 12])
prices
```

```python
# Create nut butter prices `DataFrame`
try:
    butter_prices = pd.DataFrame(prices,
                                 index=['Price'],
                                 columns=['Almond butter', 'Peanut butter', 'Cashew butter'])
except Exception as ex:
    print(ex)
```

```python
# This error exposes a "difference" between `numpy` and `pandas`;
# that is, a `numpy` array with three items is of shape (3,) 
# (three rows and an implied "one" column) but we attempt to 
# construct a `DataFrame` of size (1, 3) (1 row x 3 columns)
prices.shape
```

```python
# noinspection PyRedeclaration
butter_prices = pd.DataFrame(prices.reshape(1, 3),
                             index=['Price'],
                             columns=['Almond butter', 'Peanut butter', 'Cashew butter'])
butter_prices
```

```python
try:
    total_sales = weekly_sales.dot(butter_prices)
except ValueError as ve:
    print(f'Exception: {type(ve).__name__}: {ve}')
```

```python
weekly_sales
```

```python
# Shapes are not aligned, let's transpose
# noinspection PyRedeclaration
total_sales = prices.dot(sales_amounts.T)
total_sales
```

```python
# Create daily_sales
butter_prices
```

```python
weekly_sales
```

```python
butter_prices.shape, weekly_sales.shape
```

```python
daily_sales = butter_prices.dot(weekly_sales.T)
daily_sales
```

```python
# Doesn't work. Incompatible shapes.
try:
    weekly_sales['Total sales ($)'] = daily_sales
except Exception as ex:
    print(f'Exception: {type(ex).__name__}: {ex}')
```

```python
weekly_sales.shape, daily_sales.shape
```

```python
weekly_sales['Total sales ($)'] = daily_sales.T
weekly_sales
```
