```python
# The random numbers in `numpy` are actually pseudo-random numbers.
# One can set the random number seed to generate the a repeatable
# sequence. (Not quite random, I know.)
np.random.seed(seed=2718)
```

```python
# The following code **seems* to generate random numbers...
# But what if we wanted to share this notebooks with someone else?
random_array_4 = np.random.randint(10, size=(5, 3))
random_array_4
```

```python
random_array_4.shape
```

```python
# Best practice for generating reproducible, pseudorandom values.
# From the article "Best Practices for Using NumPy's Random Number Generators"
# found at https://blog.scientific-python.org/numpy/numpy-rng/.
rng = np.random.default_rng(seed=2718)
random_array_44 = rng.integers(10, size=(5, 3))
random_array_44
```

```python
random_array_55 = rng.random((5, 3))
random_array_55
```

```python
# The best practice seems to be to:

# - Generate a one-time 128-bit secret for the seed.
import secrets

seed = secrets.randbits(128)
seed
```

```python
# **Copy and paste** this value as the **hard-coded** random number 
# generator seed
rng = np.random.default_rng(seed=22010181874302444170331373723333688169)
```

```python
# Use `rng` whenever you need a source of random values. For example, if
# you need integer values below ten:
random_array_111 = rng.integers(10, size=(5, 3))
random_array_111
```

```python
# If you need random floating point values, use:
random_array_222 = rng.random((5, 3))
random_array_222
```
