What does the transpose do?

For a 2-dimensional array, it exchanges rows and columns. If one supplies
an additional argument, it must be a permutation of all the axes.

```python
# Create two arrays of random integers between 0 to 10
# one of size (3, 3) the other of size (3, 2)
random_array_10_3_3 = rng.integers(10, size=(3, 3))
random_array_10_3_2 = rng.integers(10, size=(3, 2))
random_array_10_3_3, random_array_10_3_2
```

```python
# Perform a dot product on the two newest arrays you created
np.dot(random_array_10_3_3, random_array_10_3_2)
```

```python
# Create two arrays of random integers between 0 to 10
# both of size (4, 3)
random_array_10_4_3_1 = rng.integers(10, size=(4, 3))
random_array_10_4_3_2 = rng.integers(10, size=(4, 3))
random_array_10_4_3_1, random_array_10_4_3_2
```

```python
# Perform a dot product on the two newest arrays you created
try:
    np.dot(random_array_10_4_3_1, random_array_10_4_3_2)
except ValueError as ve:
    print(f'{type(ve).__name__}: {ve}')
```

It doesn't work. How would you fix it?

By transposing one array or the other.

```python
# Take the latest two arrays, perform a transpose on one of them and then perform 
# a dot product on them both
np.dot(random_array_10_4_3_1, random_array_10_4_3_2.T)
```

Notice how performing a transpose allows the dot product to happen.

Why is this?

The dot product is **row by column**; consequently, the number of rows
of the first array must equal the number of columns of the second
array. The transposition of one of the arrays ensures that this
prerequisite is fulfilled.

Checking out the documentation on [`np.dot()`](https://docs.scipy.org/doc/numpy/reference/generated/numpy.dot.html) may help, as well as reading [Math is Fun's guide on the dot product](https://www.mathsisfun.com/algebra/vectors-dot-product.html).
